```python

for num in range(2, 101):
    count = 0

    for i in range(1, num + 1):
        if num % i == 0:
            count += 1

    if count == 2:
        print(num, end=" ")
```

```python

def square(n):
    for i in range(n):
        for j in range(n):
            print("*", end=" ")
        print()

square(5)
```

```python

def right_angle(n):
    for i in range(1, n + 1):
        for j in range(i):
            print("*", end=" ")
        print()

right_angle(5)
```

```python

def inverted_right(n):
    for i in range(n, 0, -1):
        for j in range(i):
            print("*", end=" ")
        print()

inverted_right(5)
```

```python

def pyramid(n):
    for i in range(1, n + 1):
        for j in range(n - i):
            print(" ", end=" ")
        for k in range(2 * i - 1):
            print("*", end=" ")
        print()

pyramid(5)
```

```python

def inverted_pyramid(n):
    for i in range(n, 0, -1):
        for j in range(n - i):
            print(" ", end=" ")
        for k in range(2 * i - 1):
            print("*", end=" ")
        print()

inverted_pyramid(5)
```

```python

def rhombus(n):
    for i in range(n):
        for j in range(n - i - 1):
            print(" ", end=" ")
        for k in range(n):
            print("*", end=" ")
        print()

rhombus(5)
```

```python

def hollow_square(n):
    for i in range(n):
        for j in range(n):
            if i == 0 or i == n-1 or j == 0 or j == n-1:
                print("*", end=" ")
            else:
                print(" ", end=" ")
        print()

hollow_square(5)
```

```python

def diamond(n):
    for i in range(1, n + 1):
        for j in range(n - i):
            print(" ", end=" ")
        for k in range(2 * i - 1):
            print("*", end=" ")
        print()

    for i in range(n - 1, 0, -1):
        for j in range(n - i):
            print(" ", end=" ")
        for k in range(2 * i - 1):
            print("*", end=" ")
        print()

diamond(5)
```

```python

def search_target(lst, target):
    for i in range(len(lst)):
        if lst[i] == target:
            return i

    return -1

numbers = [10, 20, 30, 40, 50]

target = 30

result = search_target(numbers, target)

print("Position:", result)

```

```python

def dictionary_values(d):
    values = list(d.values())

    total = sum(values)
    maximum = max(values)
    minimum = min(values)
    average = total / len(values)

    print("Total:", total)
    print("Maximum:", maximum)
    print("Minimum:", minimum)
    print("Average:", average)

d = {
    "a": 10,
    "b": 20,
    "c": 30,
    "d": 40
}

dictionary_values(d)
```

```python

def reverse_string(s):
    return s[::-1]

s = "hello"

result = reverse_string(s)

print("Original string:", s)
print("Reverse string:", result)
```
