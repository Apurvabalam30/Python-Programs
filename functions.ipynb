```python
def alphabet_count(str1):
    vowels_count = 0
    consonents_count = 0
    for ch in str1:
        if ch.isalpha():
           if ch in "aeiouAEIOU":
              vowels_count+=1
           else:
              consonents_count++1
    return[vowels_count,consonents_count]
```

```python
res = alphabet_count("python project")
print(res)
```
