```python
# factors
num = int(input("enter a number:"))
count=0
for n in range (1+num+1):
    if n%2!=0:
       count = count+1
    print(count)
```

```python
#prime numbers
number = int(input())
count = 0
for i in range(1,number+1):
    if number%i==0:
        count += 1   

if count == 2:
    print("prime")
else:
    print("not a prime")
```

```python
#perfect factor
n = int(input())
summ = 0
for i in range(1,n):
    if n%i==0:
        summ = summ+i
if summ == n:
     print("perfect")
else:
    print("not a perfect")
    
```
